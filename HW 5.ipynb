```python
# Q2
import numpy as np
import pymc as pm
import arviz as az
import matplotlib.pyplot as plt

# Generate sample data
np.random.seed(24)  # For reproducibility
n = 40
x = np.random.normal(loc=1, scale=1.5, size=n)

# 1. Bayesian inference with Normal prior for theta and Gamma prior for tau
with pm.Model() as model1:
    theta = pm.Normal('theta', mu=1, sigma=2)  # Prior for theta (Normal distribution)
    tau = pm.Gamma('tau', alpha=3, beta=2)  # Prior for tau (Gamma distribution)
    x_obs = pm.Normal('x_obs', mu=theta, sigma=1/tau, observed=x)  # Likelihood
    idata1 = pm.sample(1200, chains=2, return_inferencedata=True)  # Sampling

# 2. Bayesian inference with non-Normal prior for theta and non-Gamma prior
with pm.Model() as model2:
    theta = pm.Uniform('theta', lower=-4, upper=6)  # Prior for theta (Uniform distribution)
    tau = pm.Exponential('tau', lam=0.8)  # Prior for tau (Exponential distribution)
    x_obs = pm.Normal('x_obs', mu=theta, sigma=1/tau, observed=x)  # Likelihood
    idata2 = pm.sample(1200, chains=2, return_inferencedata=True)  # Sampling

# 3. Bayesian inference with alternative set of non-Normal and non-Gamma priors
with pm.Model() as model3:
    theta = pm.Laplace('theta', mu=1, b=2)  # Prior for theta (Laplace distribution)
    tau = pm.Beta('tau', alpha=3, beta=3)  # Prior for tau (Beta distribution)
    x_obs = pm.Normal('x_obs', mu=theta, sigma=1/tau, observed=x)  # Likelihood
    idata3 = pm.sample(1200, chains=2, return_inferencedata=True)  # Sampling

# Trace plots and summary statistics for each model
for i, idata in enumerate([idata1, idata2, idata3], start=1):
    print(f"Summary for Model {i}:")
    print(az.summary(idata))
    az.plot_trace(idata)
    plt.tight_layout()
    plt.show()

# Posterior distributions with Highest Density Interval (HDI) for each model
fig, axes = plt.subplots(3, 2, figsize=(12, 12))
az.plot_posterior(idata1, var_names=['theta', 'tau'], hdi_prob=0.94, ax=axes[0])
az.plot_posterior(idata2, var_names=['theta', 'tau'], hdi_prob=0.94, ax=axes[1])
az.plot_posterior(idata3, var_names=['theta', 'tau'], hdi_prob=0.94, ax=axes[2])
plt.tight_layout()
plt.show()
```

## Detailed Explanation of Slice Sampling

### What is Slice Sampling?
Slice sampling is a Markov Chain Monte Carlo (MCMC) method designed to sample from complex distributions where the normalization constant is unknown. Unlike Metropolis-Hastings, which requires a proposal distribution, slice sampling introduces an auxiliary variable and samples from the probability density function (PDF) by defining a "slice" under the curve. This method efficiently generates samples that form a Markov chain converging to the desired distribution.

### How Slice Sampling Works
The key idea is to sample from a region where the probability density function \( f(x) \) is above a randomly chosen threshold \( y \), avoiding direct computation of the normalizing constant.

#### Steps in the Slice Sampling Algorithm:
1. **Define the Target Function \( f(x) \)**: This is the probability density function (PDF) from which we want to draw samples.
2. **Initialization**: Choose an initial sample \( x \).
3. **Select a Vertical Slice**: Draw a random \( y \) uniformly from `[0, f(X)]`, creating a horizontal slice under the function.
4. **Determine the Interval \([a, b]\)**: Identify the range of \( x \) values.
5. **Sample a New Value \( x' \)**: Draw a new sample uniformly from this interval.
6. **Repeat**: Continue iterating the process to generate a Markov chain.

This approach effectively samples from the target distribution by iteratively refining the sampling region.

### Why Use Slice Sampling Instead of Metropolis-Hastings?
- **No need for a proposal distribution**: Unlike Metropolis-Hastings, which requires a carefully chosen proposal function, slice sampling directly samples from a valid region.
- **Improved efficiency**: It automatically adapts to the shape of the distribution, making it more effective for complex target distributions.
- **Better mixing properties**: Since it avoids rejecting samples as in Metropolis-Hastings, slice sampling often exhibits better mixing and faster convergence.

### Integrating Slice Sampling into Gibbs Sampling
When using a **Metropolis within Gibbs sampler**, if the full conditionals are only known up to a normalizing constant, slice sampling provides an efficient alternative:

1. **Full Conditional Sampling**: Instead of proposing a new state and accepting/rejecting, slice sampling directly selects from the slice defined by the current state.
2. **Defining the Slice**: The curve we sample beneath is the unnormalized full conditional density.
3. **Initial Value**: The current state of the Markov chain serves as the starting point.
4. **Iterative Sampling**:
   - Sample \( y \) uniformly from `[0, f(x_current)]`.
   - Identify the interval where \( f(x) > y \).
   - Sample uniformly from this slice.
   - Repeat for each variable in the Gibbs cycle.

This approach ensures that the stationary distribution aligns with the target posterior.

```python
### Implementation

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def slice_f_at_y(f, x, y, x_grid=np.linspace(0,1,1000)):
    """Performs one iteration of slice sampling with boundary checks."""
    x_grid_delta = x_grid[1] - x_grid[0]
    valid_indices = np.where(f(x_grid) > y)[0]
    
    if len(valid_indices) == 0:
        return x, 1  # Return initial value if no valid range found
    
    a, b = x_grid[valid_indices[[0, -1]]] + [-x_grid_delta, x_grid_delta]
    a, b = np.clip([a, b], 0, 1)  # Ensure bounds are within [0,1]
    x_ = a + stats.uniform().rvs() * (b - a)
    
    if f(x_) > y:
        return x_, 1
    elif x_ < x:
        x_l, x_r = x_, b
    else:
        x_l, x_r = a, x_
    
    return slice_f_at_y_(f, x, y, x_l, x_r, tot=2)

def slice_f_at_y_(f, x, y, x_l=0, x_r=1, tot=1):
    """Recursive function to refine the slice sampling step."""
    x_ = x_l + stats.uniform().rvs() * (x_r - x_l)
    
    if f(x_) > y:
        return x_, tot
    elif x_ < x:
        x_l = x_
    else:
        x_r = x_
    
    return slice_f_at_y_(f, x, y, x_l, x_r, tot=tot + 1)

# Define the probability density function
x_grid = np.linspace(0, 1, 1000)
f = lambda x: stats.beta(1.5, 3).pdf(x)

# Perform slice sampling
num_samples = 5000
samples = []
x_initial = 0.5  # Initial sample
for i in range(num_samples):
    y_sample = stats.uniform().rvs() * f(x_initial)
    x_new, _ = slice_f_at_y(f, x_initial, y_sample)
    samples.append(x_new)
    x_initial = x_new  # Update for the next iteration

samples = np.array(samples)  # Convert to NumPy array for plotting

# Plot the target function
plt.figure(figsize=(10, 5))
plt.plot(x_grid, f(x_grid), label='Target Distribution (Beta PDF)', color='blue')
plt.hist(samples, bins=50, density=True, alpha=0.6, label=f"{num_samples} iterations\n x values only", color='orange')
plt.legend()
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Slice Sampling of Beta(1.5, 3)")
plt.show()

```
